```python
import sys
import glob
import os
rootdir = ["/data/scratch/acw639/Medley/V1",
"/data/scratch/acw639/Medley/MedleyDB_V2/V2"]
dirs = []
for dir in rootdir:
    mix_dirs = glob.glob(os.path.join(dir, "*"))
    # remove items that are not directories
    mix_dirs = [mix_dir for mix_dir in mix_dirs if os.path.isdir(mix_dir)]
    mix_dirs = sorted(mix_dirs)  # sort
    print(len(mix_dirs))
    for mix_dir in mix_dirs:
        dirs.append(mix_dir)
    
print(len(dirs))
print(dirs[0])

```

```python
from collections import Counter
counter = {}
for dir in dirs:
    
    mix_id = os.path.basename(dir)
    track_filepaths = glob.glob(os.path.join(dir, f"{mix_id}_RAW", "*.wav"))
    num_tracks = len(track_filepaths)
    counter[mix_id] = num_tracks

    
    

```

```python
print(len(counter))
```

```python
num_tracks_list = [i for i in counter.values()]


tracks_count = Counter(num_tracks_list)

print(tracks_count)
```

```python

import random
split={}
train = []
vali = []
test = []

ctr_val = 0
ctr_train = 0
ctr_test = 0

for key, val in counter.items():
    print(val)
    if val<20 and val>4:
        if ctr_train<45:
            train.append(key)
            ctr_train = ctr_train+1
            #print("train",ctr_train)
        elif ctr_val<25:
            vali.append(key)    
            #print("vali",ctr_val)
            ctr_val = ctr_val+1
        elif ctr_test<20:
            test.append(key)    
            ctr_test = ctr_test+1
            #print("test",ctr_test)
    else:
        
        rnum = random.randint(1, 100)
        print("else",rnum)
        if rnum<60:
            train.append(key)   
            #print("train",rnum)
            
        elif rnum<80:
            vali.append(key)    
            #print("vali",rnum)
            
        else:
            test.append(key)    
            #print("test",rnum)
    # print(split["train"])
    # print(split["vali"]) 
  # print(split["test"])
split["train"] = train
split["vali"] = vali
split["test"] = test
print(len(split["test"]))
print(len(split["train"]))
print(len(split["vali"]))
```
